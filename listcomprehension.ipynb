```python
#1.Write a list comprehension to generate squares of numbers from 1 to 10.

squarenumbers = [i**2 for i in range(1,11)]
print(squarenumbers)
```

```python
#2. Create a list of even numbers between 1 and 50 using list comprehension.

evenNumbersList = [i for i in range(1,51) if i % 2 == 0 ]
print(evenNumbersList)
```

```python
#3. Convert all strings in a list to uppercase using list comprehension.

str = ['john','bob','doe']

stringUpperCase = [i.capitalize() for i in str ]
print(stringUpperCase)
```

```python
#4. Given a list of integers, create a new list that contains only the positive numbers.

list = [-1,3,-2,4,9,-5]

pList = [i for i in list if i > 0]
print(pList)
```

```python
#5. Create a list of tuples (num, num^2) for numbers 1 to 5.

tList = [(i ,i**2) for i in range(1,6)]
print(tList)



```

```python
#6. Extract all vowels from a given string using list comprehension

str = "My name is John Doe and I am in 12th standard"
vowels = [char for char in str if char in 'aeiou' ]

print(vowels)
```

```python
#7.Flatten a 2D list using list comprehension.

lt = [[1,2],[3,4],[5,6]]

flattenList = [ num for row in lt for num in row]
print(flattenList)
```

```python
#8. Replace all negative numbers in a list with 0 using list comprehension.

lti = [1,-2,3,-4]

replaceList = [0 if i< 0 else i for i in lti ]
print(replaceList)
```

```python
#9. Given a list of words, create a list of lengths of each word.

animals = ['bird' , 'pets']

animalwithlength = [len(i) for i in animals]
print(animalwithlength)
```

```python
#10. Filter out words that start with the letter 'A' or 'a'.

fruits = ['Apple' ,"Banana", "Orange" ,"ap"]


fruitsLt = [word for word in fruits if word.startswith('A') or word.startswith('a')]
print(fruitsLt)
```

```python
#11. From a list of numbers, generate a list of “even” or “odd” strings using list comprehension.
#(Like → [“even”, “odd”, “odd”, “even”…])

numbersList = [1,4,7,8,3,5,6]

new_list = [i%2 == 0 for i in numbersList]
print(new_list)
```

```python
#12. Create a list of numbers divisible by both 3 and 5 in range 1–100.

output = [i for i in range(1,101) if i % 3 ==0 and i%5 ==0]
print(output)
```

```python
#13. Write a nested list comprehension to generate a multiplication table for 1–5.

nestedList = [[i * j for j in range(1,11)] for i in range(1,6)]
print(nestedList)

```

```python
#14. Convert a dictionary’s keys into a list using list comprehension.

dict ={'name':'john','age': '30'}

cList = [i for i in dict ]

print(cList)

```

```python
#15. Extract numeric digits from a string using list comprehension.

ip = [1,'a',3, 'p']

new_ip = [i for i in ip if type(i) == int ]

print(new_ip)
```

```python
#16. Use list comprehension to remove all spaces from a string.

text = "Hello World Python"

result = "".join([char for char in text if char != " "])
print(result)
```

```python
17. #Create a list of characters that appear more than once in a string.

st ="This is the string"

op = [char for char in st if st.count(char) > 1 and char != " "]

print(op)
```

```python
#18. From a list of sentences, generate a list of all words (split using list comprehension).

li = ["This is the string This is the string This is the string "]

allwList = [ word for sentence in li for word in sentence.split()]

print(allwList)




```

```python
#19. Create a list of unique elements from a list using list comprehension + condition.

li = [1,2,5,6,8,7,7,8]

unique_elements =[]

ul = [i for i in li if  not (i in unique_elements or unique_elements.append(i))]


print(ul)


```

```python
#20. Generate all pairs (x, y) where x is from list A and y is from list B (cartesian product).

A = ['x','i','o']

B = ['y','l','j']

ot = [(i,j) for i in A for j in B]

print(ot)
```
