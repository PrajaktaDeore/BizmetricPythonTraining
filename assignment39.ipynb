```python
#39.	Create a function to accept mobile number. Mobile number should contain 10 digits. No Special character, alphabets and space.

import sys
mobilenumber = input("Please enter a mobile number")
def AcceptMobileNumber(mobilenumber):
  try:
    if mobilenumber.isdigit() and len(mobilenumber) == 10:
      print("inside if" , mobilenumber)
      return mobilenumber

    else:
      raise Exception()
      return "Invalid number"

  except ValueError:
    print("Entered mobile number is wrong")

  except:
      print(sys.exc_info())

  finally:
    print("Always go inside the loop")


AcceptMobileNumber(mobilenumber)

```
