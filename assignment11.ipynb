```python
##aks user to enter marks and make groups like distinction , first class , second class,pass, fail based on the marks
## 80 => distinction
## 70 => first class
## 50 => second class
## 35 pass
## else fail

marks = input("enter your marks")

def Get_Student_Marks(marks):
  if (not marks.isalnum() and not marks.isalpha()) or marks.isdigit():
    marks = int(marks)
    if marks > 100 or marks <0:
       output= "enter proper marks"
    elif marks >= 80 :
       output= "distinction "
    elif marks >= 70 :
       output= "second class"
    elif marks >= 50 :
       output="first class"
    elif marks == 35 :
       output="pass"
    else:
       output= "fail"
  else:
    return output
Get_Student_Marks(marks)
```
