```python
# 13. Ask user to opt for courses for master degree based on the following

# L1 = [“HR”, “Finance”, “Marketing”, “DS”]

# Based on above subject there are two different streams. 
# For example- HR is having HR core and HR analytics and 
# Marketing is having core and Marketing analytics. 
# Analytics is the optional subject and having added extra fees. 
# DS is not having analytics.

# If fees for L1 is 2 lakhs for each course core subject having the 
# same fees but analytics subject having 10% extra on 2 lakhs.

# If student opts for hostel then 2 lakhs per year is added. 
# For food monthly 2000 .

# Transportation charges 13000 per semester. 
# Calculate the total annual cost based on selected service.

# User will enter values as subject, analytics(Y/N), Hostel (Y/N), food(How many months?), Transportation(semester/annual)

subject = input("Please enter subject")
analytics = input(" Analytics (Yes/No)")
Hostel = input("Hostel(Yes/No)")
food = int(input("Food based on months"))
Transportation = input("Transportation based on semester or annual")

class Course:
     
     FEES_FOR_COURSE = 200000
     FEES_FOR_HOSTEL = 200000
     FEES_FOR_FOOD = 2000
     FEES_FOR_TRANSPORTATION = 13000
     
     def __init__(self, subject,analytics,Hostel,food,Transportation):
          self.subject = subject
          self.anlytics = analytics
          self.Hostel = Hostel
          self.food = food
          self.Transportation = Transportation

     def subjectFees(self):
          fee  =  self.FEES_FOR_COURSE

          if self.subject != 'DS'  and self.anlytics == "Y":
               fee += 0.10 * self.FEES_FOR_COURSE

          return fee

     def hostelFees(self):
        if self.Hostel == "Y":
         return self.FEES_FOR_HOSTEL
        return 0

     def foodfees(self):
       return self.food * self.FEES_FOR_FOOD

     def transportfees(self):
        if self.Transportation == "semester":
            return self.FEES_FOR_TRANSPORTATION
        elif self.Transportation == "annual":
            return self.FEES_FOR_TRANSPORTATION * 2
        else:
            return 0

     def total_cost(self):
        total = (
            self.subjectFees()
            + self.hostelFees()
            + self.foodfees()
            + self.transportfees()
        )
        print("Total Annual Cost:",total)
        return total     


obj = Course(subject ,analytics, Hostel, food, Transportation)
obj.subjectFees()
obj.hostelFees()
obj.foodfees()
obj.transportfees()
obj.total_cost()
```
