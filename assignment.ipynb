```python
#2424.	Ask user to enter the name and DOB and generate the password based on first name 4 characters and  @ddmm.
#  For example name = SURESH and DOB = 05-05-1978 then password will be SURE@0505 

class PasswordG:

    def __init__(self, name, dob):
        self.name = name
        self.dob = dob

    def generate_password(self):
        try:
            if not self.name.isalpha():
                raise ValueError("Name should contain alphabets")

            parts = self.dob.split("-")
            if len(parts) != 3:
                raise ValueError("DOB must be in dd-mm-yyyy")

            dd, mm, yyyy = parts

            if not (dd.isdigit() and mm.isdigit() and yyyy.isdigit()):
                raise ValueError("DOB should contain only numbers.")

            fpart = self.name[:4].upper()

            password = fpart + "@" + dd + mm

            return password

        except Exception as e:
            return f"Error: {e}"


name = input("Enter your name ")
dob = input("Enter your DOB ")

obj = PasswordG(name, dob)
print("Generated Password:", obj.generate_password())
```

```python
#25. Ask user to enter the name and DOB and generate the password based on first name 4 characters and  last 4 digits of DOB or year @yyyy. 
# For example name = SURESH and DOB = 05-05-1978 then password will be SURE@1978

class Generator:

    def __init__(self, name, dob):
        self.name = name
        self.dob = dob

    def generate_pw(self):
        first_part = self.name[:4].upper()
        year = self.dob.split("-")[-1]
        password = first_part + "@" + year
        return password

name = input("Enter name")
dob = input("Enter your DOB")

obj = Generator(name, dob)
print("Generated Password:", obj.generate_pw())
```

```python
# 35.	List2 = [2,70,'work', para, 2.5, [1,2,3], (1,2), {1,2}, {1:'a', 2:'b'}, 3,10,302.5]
# Multiply each and every element by 2 and display the answer

List2 = [2, 70, 'work', 'para', 2.5, [1,2,3], (1,2), {1,2}, {1:'a', 2:'b'}, 3, 10, 302.5]

output = []

for item in List2:
    if isinstance(item, (int, float, str, list, tuple)):
        output.append(item * 2)
    
    elif isinstance(item, set):
        output.append({x * 2 for x in item})
    
    elif isinstance(item, dict):
        output.append({k * 2: v * 2 for k, v in item.items()})
    
    else:
        output.append(item)

print("Result:", output)

```

```python
# 48.	Create a empty dictionary and ask user to enter values as name, DOB, mobile number 
# add all the details in dictionary with customer number as 1 for first time. If user try to enter another value, 
# then number should increase as 2 with new details and previous values should not change.
# For example:
# {}
# {1:{name : "Sachin", "DOB": "21-06-1965" , "mobile": "1234123423"}}

# {1:{name : "Sachine", "DOB": "21-06-1965" , "mobile": "1234123423"},
# 2: {name : "Sumedh", "DOB": "02-02-2002" , "mobile": "1234123433"}}

cstmer = {} 
customer_id = 1

while True:
    name = input("Enter Name: ")
    dob = input("Enter DOB (dd-mm-yyyy): ")
    mobile = input("Enter Mobile Number: ")

    # Add details inside dictionary
    cstmer[customer_id] = {
        "name": name,
        "DOB": dob,
        "mobile": mobile
    }
    print(cstmer)

    customer_id += 1  # increase customer number

    choice = input("\nDo you want to add another customer? (yes/no): ")
    if choice.lower() != "yes":
        break


with open("cust_info.txt", "w") as file:
    for cid, details in cstmer.items():
        file.write(f"Customer ID: {cid}\n")
        for key, value in details.items():
            file.write(f"{key}: {value}\n")
        file.write("-" * 30 + "\n")

print("Customer data saved successfully in cust_info.txt")


```

```python
# 44.	Create a table  cust_info as sr_no, name, DOB, mobile. 
# Ask user to enter the information from python code. Validate all fields and after validation insert records in the table.

import pyodbc
import re


conn = pyodbc.connect(
            "DRIVER={SQL Server};"
            "SERVER=SUDARSHAN\\SQLEXPRESS;"
            "DATABASE=Customer;"
            "Trusted_Connection=yes;"
        )

cursor = conn.cursor()


def validate_data(name, dob, mobile):
    if not name.isalpha():
        raise ValueError("Name must contain only alphabets.")

    if not re.match(r"\d{2}-\d{2}-\d{4}$", dob):
        raise ValueError("DOB must be in dd-mm-yyyy format.")

    if not (mobile.isdigit() and len(mobile) == 10):
        raise ValueError("Mobile must contain exactly 10 digits.")

    return True


try:
    name = input("Enter Name: ")
    dob = input("Enter DOB (dd-mm-yyyy): ")
    mobile = input("Enter Mobile Number: ")

    validate_data(name, dob, mobile)

    cursor.execute("INSERT INTO cust_info (name, DOB, mobile) VALUES (?, ?, ?)",
                   (name, dob, mobile))

    conn.commit()

    print("Record inserted successfully.")

except Exception as e:
    print("Error:", e)

finally:
    conn.close()
```

```python
# 46.	Create a function to calculate age till now
from datetime import date

def calculate_age(year, month, day):
    today = date.today()
    birth_date = date(year, month, day)

    age = today.year - birth_date.year

    if (today.month, today.day) < (birth_date.month, birth_date.day):
        age -= 1

    return age

age = calculate_age(2000, 5, 10)
print("Current Age:", age)
```

```python
# 47.Create a function to check age eligibility for given customer based on DOB. Function will take two input DOB and ELIGIBILITY age.

from datetime import date

def eligibility_chk():
    year = int(input("Enter birth year: "))
    month = int(input("Enter birth month: "))
    day = int(input("Enter birth day: "))
    eligibility_age = int(input("Enter eligibility age: "))

    today = date.today()
    birthdate = date(year, month, day)

    age = today.year - birthdate.year

    if (today.month, today.day) < (birthdate.month, birthdate.day):
        age -= 1

    if age >= eligibility_age:
        print("Customer is Eligible")
    else:
        print("Customer is Not Eligible")


eligibility_chk()
```

```python
#50.Write a code to generate factorial of the number  For example: factorial of 5 = 5! = 5*4*3*2*1

def fact(number):
    if number < 0:
        print(" negative numbers")
        return
    
    result = 1
    for i in range(1, number + 1):
        result *= i
    
    return result


num = 5
print(f"Factorial of {num} is:", fact(num))
```
