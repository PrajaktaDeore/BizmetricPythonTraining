```python
# 40.	Create a function to generate auto-password based on specific person details. Ask user to enter name, DOB.
# And password must be First name 4 characters and year of birth.


def generate_password():

    name = input("Please enter full name: ").strip()
    dob = input("Enter DOB  format is (DD-MM-YYYY): ").strip()

    firstName = name.split()[0]
    first_four_characters = firstName[:4]
    birthyear = dob.split("-")[2]

    password = first_four_characters + birthyear

    print("Password:", password)

generate_password()
```
