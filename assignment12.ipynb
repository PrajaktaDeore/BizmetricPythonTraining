```python
# 12.	Ask user to enter information about salary of the employee per year and rating received as A, B, C, D
# If salary upto 5 lpa then increament based on ratings are  A = 16% , B=12%, C= 10%, D=6%
# If salary upto 10 lpa then increament based on ratings are  A = 14% , B=10%, C= 8%, D=6%
# If salary upto 15 lpa then increament based on ratings are  A = 8% , B=6%, C= 4%, D = no increment
# If salary upto 23 lpa then increament based on ratings are  A = 7% , B=5%, C= 4%, D=no increment

salary = input("please enter a salary")
rating = input("please enter a rating as A, B,C,D")


def SalaryCalculation(salary,rating):
 salary = int(salary)
 count = 0
 if salary <= 5:
    if rating == 'A':
        count = 16

    elif rating == 'B':
        count =12

    elif  rating == 'c':
        count =10

    elif rating =='D':
        count =6


 if salary <=10:
   if rating == 'A':
        count = 14

   elif rating == 'B':
        count =10

   elif  rating == 'c':
        count =8

   elif rating =='D':
        count =6

 if salary <=15:
     if rating == 'A':
        count = 8

     elif rating == 'B':
        count =6

     elif  rating == 'c':
        count =4

     elif rating =='D':
        count =0


 if salary <=23:
     if rating == 'A':
        count = 7

     elif rating == 'B':
        count = 5

     elif  rating == 'c':
        count = 4

     elif rating =='D':
        count =0

 new_salary = salary + (salary * count / 100)
 print(new_salary ,count)

 return count, new_salary

SalaryCalculation(salary,rating)

```
