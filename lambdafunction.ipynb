```python
#1.Write a lambda to add two numbers.

addition =  lambda x,y : x + y

print(addition(3, 4))
```

```python
#2.Create a lambda to check if a number is even.

evenNumber = lambda num : num % 2 ==0

print(evenNumber(4))
```

```python
#3.Write a lambda to get the last character of a string.


char = lambda abc: abc[-1]

print(char("Hello"))
```

```python
#4.Use lambda with map() to square every number in a list.

n =[1,3,4,5]
sq_numbers = list(map(lambda x: x * x, n))

print(sq_numbers)
```

```python
#5.Use lambda with filter() to get only odd numbers from a list.

list1 = [1,2,3,4,5,6,7]
filtered_list = list(filter(lambda i: i % 2 != 0,list1))

print(filtered_list)
```

```python
#6. Use sorted() + lambda to sort a list of tuples by second value.

ltT = [(1,4), (3,2)]
sortedList = sorted(ltT, key=lambda j : j[1])

print(sortedList)
```

```python
#7. Create a lambda to check if a string is a palindrome.

output = lambda string1 :  string1 == string1[::-1]

print(output("MADAM"))
print(output("abc"))
```

```python
#8.Use lambda to find maximum of three numbers.


max_num = lambda x,y,z : x if (x >= y or x>=z) else (y if y>=z else z)

maximum = lambda a,b,c : max(a,b,c)

print(max_num(1,2,3))
print(maximum(1,2,3))
```

```python
#9.Write a lambda to reverse a string.

rev_string = lambda x : x[::-1]

print(rev_string("Hello"))
```

```python
#11.Use lambda with filter() to remove empty strings from a list.
strt = ["ab", " " ,"cd"]

rem_em_strt = list(filter(lambda x: x != " " ,strt))

print(rem_em_strt)
```

```python
#12. Use lambda to compute factorial using reduce() (yeah, that one-liner madness).

from functools import reduce
fact = lambda x: reduce(lambda a, b: a * b, range(1, x + 1), 1)
print(fact(4))
```

```python
#20.Write a lambda to remove duplicates from a list using filter + set.

n = [1, 3, 3, 2, 6, 6, 5]

unique_n = list(filter(lambda x: x in set(n), n))

print(list(set(unique_n)))
```

```python
#21.Use lambda with reduce() to find the product of all numbers in a list.

from functools import reduce

n = [1, 2, 3, 4, 5]
nproduct = reduce(lambda x, y: x * y, n)
print(nproduct)
```

```python
#25.Write a lambda that returns the square if number is even, cube if odd.

out = lambda x :  x**2 if (x % 2 == 0) else x**3

print(out(3))
print(out(4))
```

```python
#26.Use lambda with map to convert Celsius to Fahrenheit.

cel = [0, 10, 20, 30, 40]

fahren = list(map(lambda c: (c * 9/5) + 32, cel))

print(fahren)
```

```python
#27.Write a lambda to check if two strings are anagrams.

is_anag = lambda s1, s2: sorted(s1) == sorted(s2)

print(is_anag("listen", "silent"))
print(is_anag("Hi", "Hello"))

```

```python
#28.Use lambda to extract only numeric values from a mixed list.

lt = ['a',2,'b',3]

numeric_values = list(filter(lambda x: isinstance(x, (int)), lt))

print(numeric_values)
```

```python
#29.Use lambda inside any() to check if any list element is negative.

n = [5, 10, -3, 8, 2]

has_negative = any(map(lambda x: x < 0, n))

print(has_negative)
```

```python
#30. Use lambda to generate a function that multiplies any number by n

np = lambda x : lambda n : x * n

num1 = np(8)
num2 = np(5)
print(num1(4))
print(num2(5))
```
